### Objective:
Train 4 distinct machine learning & deep learning architectures on the benchmark dataset:
1. **Baseline Linear Model**: Ridge Regression with StandardScaler fitted strictly on training data
2. **Random Forest**: 100 decision trees + ensemble variance estimation for calibrated predictive uncertainty
3. **XGBoost Regressor**: Gradient-boosted trees with tree subsampling
4. **1D CNN Sequence Model**: PyTorch convolutional network discovering sequence motifs directly from one-hot matrices


```python
import sys
sys.path.append("..")
import os
import joblib
import json
import torch
from backend.app.training.train import run_training_pipeline

# Check if pre-trained models already exist
SAVED_DIR = os.path.join("..", "backend", "saved_models")
if os.path.exists(os.path.join(SAVED_DIR, "metrics.json")):
    with open(os.path.join(SAVED_DIR, "metrics.json")) as f:
        metrics = json.load(f)
    print("Models are already trained!")
    print("Best Model Identified:", metrics.get("best_model"))
    print("\nRandom Split Performance Summary:")
    for model_name, m_dict in metrics.get("random_split", {}).items():
        print(f"  {model_name:18} -> Pearson r: {m_dict['pearson']:.4f}, Spearman rho: {m_dict['spearman']:.4f}, RMSE: {m_dict['rmse']:.4f}")
else:
    print("Running training pipeline...")
    run_training_pipeline()
```
